```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("D:\\PLACEMENT PREDICTION\\data\\titanic.csv")
```

```python
print("1. Missing Age values:", df["Age"].isnull().sum())
```

```python
print("2. Number of passengers survived:", (df["Survived"] == 1).sum())
```

```python
print("3. Number of people dead:", (df["Survived"] == 0).sum())
```

```python
df = df.dropna(subset=["Age"])
print("\nRows after removing missing Age values:", len(df))
```

```python
print("\n5. Names and Ages of Survived People")
print(df[df["Survived"] == 1][["Name", "Age"]])
```

```python
print("\n6. Numeric DataFrame")
numeric_df = df.select_dtypes(include="number")
print(numeric_df)
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df["Age"], kde=True)
plt.title("Distribution Plot of Age")
plt.show()
```

```python
plt.figure(figsize=(5,4))
sns.countplot(x="Survived", data=df)
plt.title("Count Plot of Survived")
plt.show()
```

```python
survived_gender = df[df["Survived"] == 1]["Sex"].value_counts()

print("\n9. Survived Males and Females")
print(survived_gender)

if survived_gender["female"] > survived_gender["male"]:
    print("Conclusion: More females survived than males.")
else:
    print("Conclusion: More males survived than females.")
```

```python
plt.figure(figsize=(6,4))
sns.countplot(x="Pclass", data=df[df["Survived"] == 1])
plt.title("Survived People in Each Passenger Class")
plt.show()

survived_pclass = df[df["Survived"] == 1]["Pclass"].value_counts()

print("\n10. Survived People by Pclass")
print(survived_pclass)

if survived_pclass[1] > survived_pclass[2] and survived_pclass[1] > survived_pclass[3]:
    print("Conclusion: More first-class passengers survived than second and third class passengers.")
```
